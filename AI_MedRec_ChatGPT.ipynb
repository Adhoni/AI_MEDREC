```python
def get_medication_from_handwritten_notes(text, example):
    # System-level instructions
    system_prompt = """You will be given an example text. Your role is to identify and extract the medication list from the given text and return the identified 1. tablet name, 2. administration, 3. dosage which is expressed as mg, mcg, ml or number of caps, and 4. frequency as illustrated in the example response. Give response in the exact order as stated. The example response is a list of tuples with all values in string. Return None if any of the value is not present."""

    # User's prompt (handwritten note + example)
    user_prompt = f"\nText:\n{text}\n\n{example}"
    combined_prompt = f"{system_prompt}\n\n{user_prompt}"

    # print("Combined prompt:\n")
    # print(combined_prompt)
    response = client.chat.completions.create(
        model='gpt-4o',
        messages=[
            {'role': 'user', 'content': combined_prompt}
        ],
        temperature=0,
    )
    # Extract the response text
    response_text = response.choices[0].message.content.strip()
    return response_text
```
