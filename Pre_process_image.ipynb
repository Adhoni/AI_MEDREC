```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
```

```python

def preprocess_image(image_path):
    # Read the image
    image = cv2.imread(image_path)

    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Apply Gaussian blur
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # Perform adaptive thresholding
    thresh = cv2.adaptiveThreshold(blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 11, 2)

    # Perform morphological operations to remove noise
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
    opened = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

    return opened

# Example usage:
image_path = '/content/drive/MyDrive/Colab Notebooks/AP/Pre-process-pill-images/CIPLA-159.png'
processed_image = preprocess_image(image_path)

# Display the preprocessed image
cv2_imshow(processed_image)
```

```python
def denoise_image(image):
    # Apply non-local means denoising
    denoised_image = cv2.fastNlMeansDenoisingColored(image, None, 10, 10, 7, 21)

    return denoised_image

# Read the colored image
image_4 = cv2.imread('/content/drive/MyDrive/Colab Notebooks/AP/Pre-process-pill-images/H-502-hydroxyzine-hydrochloride.png')

# Denoise the image
denoised_image = denoise_image(image_4)

# Display the original and denoised images
cv2_imshow(image_4)
print("Denoised below")
cv2_imshow(denoised_image)
```

```python
def sharpen_image(image):
    # Convert the image to float32
    image_float = np.float32(image)

    # Create a blurred version of the image
    blurred = cv2.GaussianBlur(image_float, (0,0), 3)

    # Calculate the difference between the original and blurred image
    sharpened = cv2.addWeighted(image_float, 1.5, blurred, -0.5, 0)

    # Convert back to uint8
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

    return sharpened
```

```python
# Read the colored image
image = cv2.imread('/content/drive/MyDrive/Colab Notebooks/AP/Pre-process-pill-images/CIPLA-159.png')

# Sharpen the image
sharpened_image = sharpen_image(image)

# Display the original and sharpened images
cv2_imshow(image)
print("Below sharpened")
cv2_imshow(sharpened_image)
```
