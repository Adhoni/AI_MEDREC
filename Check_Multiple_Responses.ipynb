```python
import google.generativeai as genai
import os
import PIL.Image
```

```python
def get_response(question):

    safety_settings = [
        {
            "category": "HARM_CATEGORY_DANGEROUS",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_HARASSMENT",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_HATE_SPEECH",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_SEXUALLY_EXPLICIT",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_DANGEROUS_CONTENT",
            "threshold": "BLOCK_NONE",
        },
    ]

    config = genai.GenerationConfig(
      candidate_count=2,
      temperature=0.1
    )

    genai.configure(api_key="AIzaSyCQxqXyDj7jcqxhNFEtIuii3GlrHyItwwk")
    model = genai.GenerativeModel('gemini-pro', safety_settings=safety_settings, generation_config=config)
    response = model.generate_content(question)

    candidates = response.candidates  # Get all candidates
    print("\n")
    print("Length of candidates -> ", len(candidates))
    print("All candidates -> ", candidates)
    print("\n")
    candidates = response.candidates  # Get all candidates
    all_answers = [candidate.content for candidate in candidates]
    print("All answers -> ", all_answers)
    return response.text
    # return temp_text
```

```python
question_1 = "What is Cucumber?"
response = get_response(question_1)
print("\n\nGemini response \n", response)
```

```python
def ocr_pill(image):
    print(image)
    # system_prompt = """Extract the text. Give only the text and nothing else.\n"""
    system_prompt = """Extract the text. The text may be curved so extract it correctly. Give only the text and nothing else.\n"""
    contents = [image, system_prompt]
    safety_settings = [
        {
            "category": "HARM_CATEGORY_DANGEROUS",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_HARASSMENT",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_HATE_SPEECH",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_SEXUALLY_EXPLICIT",
            "threshold": "BLOCK_NONE",
        },
        {
            "category": "HARM_CATEGORY_DANGEROUS_CONTENT",
            "threshold": "BLOCK_NONE",
        },
    ]

    config = genai.GenerationConfig(
      candidate_count=1,
      temperature=0.1
    )

    genai.configure(api_key="AIzaSyCQxqXyDj7jcqxhNFEtIuii3GlrHyItwwk")
    model = genai.GenerativeModel('gemini-pro-vision',safety_settings=safety_settings, generation_config=config)
    response = model.generate_content(contents)

    candidates = response.candidates  # Get all candidates
    print("\n")
    print(type(candidates))
    print(len(candidates))
    print("All candidates ", candidates)
    print("\n")
    # temp_text = candidates[0].content  # Access text from first candidate
    print("temp_text ", response.parts)
    candidates = response.candidates  # Get all candidates
    all_answers = [candidate.content for candidate in candidates]
    print("All answers -> ", all_answers)
    return response.text
```
